```python
import pandas as pd
from TripValidator import *
from TripProcessor import *

file_path = "../../input/realtime/processed/Traveler_trip_details.csv"
df = pd.read_csv(file_path)

pre_validator = PreTripProcessorValidator(df)
pre_validator.run_all_validations()
print("Pre-validation Errors:", pre_validator.errors)

# 3. PROCESS THE DATA FIRST
processor = TripProcessor(df)  # you must have a TravelerProcessor class
processed_df = processor.run_all_processors().get_processed_data()

# 4. NOW validate the processed data
post_validator = PostTripProcessorValidator(processed_df)
post_validator.run_all_validations()

# 查看验证结果
print("Post-validation Errors:", post_validator.errors)
```
